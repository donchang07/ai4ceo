# API key를 테스트 

```python
import os
from dotenv import load_dotenv

# LangChain Chat Models
from langchain_openai import ChatOpenAI
from langchain_anthropic import ChatAnthropic
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_perplexity import ChatPerplexity
from langchain_upstage import ChatUpstage
from langchain_community.tools.tavily_search import TavilySearchResults
from langchain_core.runnables import RunnableLambda

# .env 파일 로드 (기존 값 덮어쓰기)
load_dotenv(override=True)

# 환경 변수에서 API 키 가져오기
openai_api_key = os.getenv('OPENAI_API_KEY')
tavily_api_key = os.getenv('TAVILY_API_KEY')
google_api_key = os.getenv('GOOGLE_API_KEY')
anthropic_api_key = os.getenv('ANTHROPIC_API_KEY')
perplexity_api_key = os.getenv('PERPLEXITY_API_KEY')
upstage_api_key = os.getenv('UPSTAGE_API_KEY')
deepl_api_key = os.getenv('DEEPL_API_KEY')
```

```python
def _mask(k):
    if not k:
        return "(missing)"
    return k[:6] + "..." + k[-4:] if len(k) > 14 else "(set)"

for label, key in [
    ("OPENAI_API_KEY", openai_api_key),
    ("TAVILY_API_KEY", tavily_api_key),
    ("GOOGLE_API_KEY", google_api_key),
    ("ANTHROPIC_API_KEY", anthropic_api_key),
    ("PERPLEXITY_API_KEY", perplexity_api_key),
    ("UPSTAGE_API_KEY", upstage_api_key),
    ("DEEPL_API_KEY", deepl_api_key),
]:
    print(f"{label}: {_mask(key)}")
```

```python
# OpenAI API 테스트 (LangChain)
try:
    llm = ChatOpenAI(model="gpt-5.2", max_tokens=500)
    response = llm.invoke([
        {"role": "system", "content": "당신은 친절한 AI Assistance 입니다."},
        {"role": "user", "content": "say this is test"},
    ])
    print("OpenAI API 성공")
except Exception as e:
    print(f"OpenAI API 실패: {e}")
```

```python
# Google Gemini API 테스트 (LangChain)
try:
    llm = ChatGoogleGenerativeAI(model="gemini-3-flash-preview", google_api_key=google_api_key)
    response = llm.invoke("say this is test")
    print("Google Gemini API 성공")
except Exception as e:
    print(f"Google Gemini API 실패: {e}")
```

```python
# Anthropic API 테스트 (LangChain)
try:
    llm = ChatAnthropic(model="claude-sonnet-4-6", api_key=anthropic_api_key, max_tokens=500)
    response = llm.invoke("say this is test")
    print("Anthropic API 성공")
    print(f"응답: {response.content}")
except Exception as e:
    print(f"Anthropic API 실패: {e}")
```

```python
# Perplexity API 테스트 (LangChain)
try:
    llm = ChatPerplexity(api_key=perplexity_api_key, model="sonar-pro")
    response = llm.invoke("This is a test message")
    print("Perplexity API 성공")
except Exception as e:
    print(f"Perplexity API 실패: {e}")
```
